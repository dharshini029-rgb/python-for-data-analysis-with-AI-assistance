task 1: resample daily sales into weekly aggregate sales


```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is in datetime format (already done in previous cells, but good practice to ensure)
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])

# Set 'Order Date' as index
df_indexed = df_superstore.set_index('Order Date')

# Resample 'Sales' weekly ('W') using sum aggregation
weekly_sales = df_indexed['Sales'].resample('W').sum()

# Plot the time chart
plt.figure(figsize=(12, 6))
weekly_sales.plot()
plt.title('Weekly Total Sales (Sum)')
plt.xlabel('Order Date')
plt.ylabel('Total Sales')
plt.grid(True)
plt.tight_layout()
plt.show()
```

task 2: calculate and plot 7-day moving average of sales

```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is datetime and set as index (if not already for this df)
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
df_daily_sales = df_superstore.set_index('Order Date').resample('D')['Sales'].sum()

# Calculate the 7-day rolling mean
df_daily_sales_rolling_mean = df_daily_sales.rolling(window=7).mean()

# Plot original daily sales and the 7-day rolling mean
plt.figure(figsize=(15, 7))
plt.plot(df_daily_sales, label='Daily Sales', alpha=0.7)
plt.plot(df_daily_sales_rolling_mean, label='7-Day Rolling Mean', color='red')
plt.title('Daily Sales vs. 7-Day Rolling Mean')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

task 3: calculate period-over-period daily growth using. shift()

```python
# Ensure 'Order Date' is datetime and set as index for daily sales
df_daily_sales = df_superstore.set_index('Order Date').resample('D')['Sales'].sum()

# Create a lag-1 column 'sales_prev_day'
df_daily_sales = pd.DataFrame(df_daily_sales)
df_daily_sales['sales_prev_day'] = df_daily_sales['Sales'].shift(1)

# Calculate daily percentage change in sales
df_daily_sales['daily_percentage_change'] = ((df_daily_sales['Sales'] - df_daily_sales['sales_prev_day']) / df_daily_sales['sales_prev_day']) * 100

# Display the results
display(df_daily_sales.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Calculate Profit Margin (Profit / Sales)
# Handle potential division by zero by setting Profit_Margin to 0 if Sales is 0
df_superstore['Profit_Margin'] = np.where(
    df_superstore['Sales'] != 0,
    (df_superstore['Profit'] / df_superstore['Sales']) * 100,
    0
)

plt.figure(figsize=(10, 6))
sns.histplot(
    df_superstore['Profit_Margin'],
    bins=50, # Adjust number of bins as needed
    kde=True, # Display Kernel Density Estimate
    color='skyblue', # Set color for the histogram
    edgecolor='black' # Set edge color for bins
)

plt.title('Distribution of Order Profit Margin')
plt.xlabel('Profit Margin (%)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

task 2: Matplotib 3D surface plot of synthetic sales-profit surface

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
import pandas as pd

# Ensure df_superstore is available (assuming it's already loaded in previous cells)

# Define the range for Sales (x) and Profit (y)
# Using the min/max from the actual data for realistic scale
x_sales = np.linspace(df_superstore['Sales'].min(), df_superstore['Sales'].max(), 100)
y_profit = np.linspace(df_superstore['Profit'].min(), df_superstore['Profit'].max(), 100)

# Create a meshgrid
X, Y = np.meshgrid(x_sales, y_profit)

# Define a synthetic Z (Quantity) surface as a function of X (Sales) and Y (Profit)
# For demonstration, let's use a simple quadratic function that might resemble some real-world trends.
# You might adjust this function based on actual domain knowledge.
# Adding a small constant to avoid division by zero or negative values if needed
Z = (0.0001 * X**2) + (0.001 * Y) + 5 # Example: Quantity increases quadratically with Sales and linearly with Profit

fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surface = ax.plot_surface(
    X, Y, Z,
    cmap='viridis', # Choose a colormap
    edgecolor='none' # Remove edges for a smoother look
)

# Set labels
ax.set_xlabel('Sales')
ax.set_ylabel('Profit')
ax.set_zlabel('Quantity (Synthetic)')
ax.set_title('3D Surface Plot of Synthetic Quantity vs. Sales and Profit')

# Add a color bar to indicate Z values
fig.colorbar(surface, shrink=0.5, aspect=10, label='Quantity (Synthetic)')

plt.tight_layout()
plt.show()
```

```python
import dash
from dash import dcc, html, dash_table
from dash.dependencies import Input, Output
import pandas as pd

# Initialize the Dash app
app_table_summary = dash.Dash(__name__)

# --- Data Preparation ---
# Get unique regions for dropdown options, including an 'All Regions' option
available_regions_table = ['All Regions'] + sorted(df_superstore['Region'].unique().tolist())

# Define the app layout
app_table_summary.layout = html.Div([
    html.H1("Superstore Sales Data & Summary", style={'textAlign': 'center', 'marginBottom': '30px'}),

    html.Div([
        html.Label("Select Region:", style={'marginRight': '10px'}),
        dcc.Dropdown(
            id='region-selector-table',
            options=[{'label': region, 'value': region} for region in available_regions_table],
            value='All Regions', # Default to 'All Regions'
            multi=False, # Allow only single selection
            clearable=False,
            placeholder="Select a Region"
        ),
    ], style={'width': '50%', 'margin': 'auto', 'padding': '20px'}),

    html.Div([
        html.H3(id='total-sales-summary', style={'textAlign': 'center', 'color': '#007BFF'}),
    ], style={'marginTop': '20px', 'marginBottom': '20px'}),

    html.Div([
        dash_table.DataTable(
            id='filtered-data-table',
            columns=[{"name": i, "id": i} for i in df_superstore.columns],
            page_size=10, # Display 10 rows per page
            style_table={'overflowX': 'auto'},
            style_cell={'textAlign': 'left', 'padding': '5px'},
            style_header={
                'backgroundColor': 'rgb(230, 230, 230)',
                'fontWeight': 'bold'
            }
        )
    ], style={'width': '90%', 'margin': 'auto'})
])

# Define the callback to update the table and summary based on region selection
@app_table_summary.callback(
    [Output('filtered-data-table', 'data'),
     Output('total-sales-summary', 'children')],
    [Input('region-selector-table', 'value')]
)
def update_table_and_summary(selected_region):
    if selected_region == 'All Regions':
        filtered_df = df_superstore
    else:
        filtered_df = df_superstore[df_superstore['Region'] == selected_region]

    # Calculate total sales for the selected region
    total_sales = filtered_df['Sales'].sum()
    summary_text = f"Total Sales for {selected_region}: ${total_sales:,.2f}"

    # Return data for the DataTable and the summary text
    return filtered_df.to_dict('records'), summary_text

# Run the app (for Colab, use mode='inline')
app_table_summary.run(mode='inline')
```

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd

# Initialize the Dash app
app = dash.Dash(__name__)

# --- Data Preparation ---
# Get unique regions and categories for dropdown options
available_regions = df_superstore['Region'].unique()
available_categories = df_superstore['Category'].unique()

# Get min/max sales for the slider
min_sales = df_superstore['Sales'].min()
max_sales = df_superstore['Sales'].max()

# Define the app layout
app.layout = html.Div([
    html.H1("Superstore Sales Dashboard", style={'textAlign': 'center', 'marginBottom': '30px'}),

    html.Div([ # Main container with sidebar and content
        # Left Sidebar for Controls
        html.Div([
            html.H3("Filters", style={'marginBottom': '20px'}),

            html.Label("Select Region:", style={'marginTop': '10px'}),
            dcc.Dropdown(
                id='region-filter',
                options=[{'label': region, 'value': region} for region in available_regions],
                value=[available_regions[0]], # Default to first region, allow multiple
                multi=True, # Allow multiple selections
                placeholder="Select Region(s)"
            ),

            html.Label("Select Category:", style={'marginTop': '20px'}),
            dcc.Dropdown(
                id='category-filter',
                options=[{'label': category, 'value': category} for category in available_categories],
                value=[available_categories[0]], # Default to first category, allow multiple
                multi=True, # Allow multiple selections
                placeholder="Select Category(s)"
            ),

            html.Label("Sales Range:", style={'marginTop': '20px'}),
            dcc.RangeSlider(
                id='sales-range-slider',
                min=min_sales,
                max=max_sales,
                value=[min_sales, max_sales], # Default to full range
                marks={str(int(min_sales)): str(int(min_sales)),
                       str(int(max_sales)): str(int(max_sales))},
                tooltip={'always_visible': True, 'placement': 'bottom'}
            ),
        ], style={'width': '25%', 'padding': '20px', 'boxShadow': '2px 2px 5px #cccccc', 'marginRight': '20px'}),

        # Right Content Area for Graph
        html.Div([
            dcc.Graph(id='sales-by-category-graph')
        ], style={'width': '75%', 'padding': '20px'})

    ], style={'display': 'flex'}) # End Main container
])

# Define the callback to update the graph based on dropdown and slider selections
@app.callback(
    Output('sales-by-category-graph', 'figure'),
    [Input('region-filter', 'value'),
     Input('category-filter', 'value'),
     Input('sales-range-slider', 'value')]
)
def update_graph(selected_regions, selected_categories, sales_range):
    # Ensure inputs are lists for consistent filtering, handle None for cleared dropdowns
    if not selected_regions:
        selected_regions = [] # If no regions selected, filter out everything initially
    if not isinstance(selected_regions, list):
        selected_regions = [selected_regions]

    if not selected_categories:
        selected_categories = [] # If no categories selected, filter out everything initially
    if not isinstance(selected_categories, list):
        selected_categories = [selected_categories]

    # Filter by Region
    filtered_df = df_superstore[df_superstore['Region'].isin(selected_regions)]

    # Filter by Category
    filtered_df = filtered_df[filtered_df['Category'].isin(selected_categories)]

    # Filter by Sales Range
    filtered_df = filtered_df[(filtered_df['Sales'] >= sales_range[0]) & (filtered_df['Sales'] <= sales_range[1])]

    # Group by Category and sum Sales for the filtered data
    df_category_sales = filtered_df.groupby('Category')['Sales'].sum().reset_index()

    # If no data after filtering, return an empty figure or a message
    if df_category_sales.empty:
        return px.bar(title="No data available for the selected filters.")

    fig = px.bar(
        df_category_sales,
        x='Category',
        y='Sales',
        title=f'Total Sales by Category for Selected Filters',
        labels={'Category': 'Product Category', 'Sales': 'Total Sales'},
        color='Category', # Color bars by category
        template='plotly_white'
    )
    fig.update_layout(xaxis_title_text='Category', yaxis_title_text='Total Sales ($)')
    return fig

# Run the app (for Colab, use mode='inline')
app.run(mode='inline')
```

### Task 10: Multi-component KPI Card and Category Pie Chart Layout (Dash App)

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd

# Initialize the Dash app for KPI cards and pie chart
app_kpi_pie = dash.Dash(__name__)

# --- Data Preparation ---
# Get unique regions for dropdown options
available_regions_kpi = df_superstore['Region'].unique()

# Define the app layout
app_kpi_pie.layout = html.Div([
    html.H1("Superstore Performance Dashboard", style={'textAlign': 'center', 'marginBottom': '30px'}),

    html.Div([ # Filter section
        html.Label("Select Region:", style={'marginRight': '10px'}),
        dcc.Dropdown(
            id='region-dropdown-kpi',
            options=[{'label': region, 'value': region} for region in available_regions_kpi],
            value='All Regions', # Default to all regions
            multi=False,
            clearable=False,
            placeholder="Select a Region"
        ),
    ], style={'width': '50%', 'margin': 'auto', 'padding': '20px'}),

    html.Div([ # KPI Cards section
        html.Div(id='kpi-cards-output', style={'display': 'flex', 'justifyContent': 'space-around', 'marginTop': '20px'})
    ]),

    html.Div([ # Pie Chart section
        dcc.Graph(id='category-sales-pie-chart')
    ], style={'width': '80%', 'margin': 'auto', 'marginTop': '30px'})
])

# Callback to update KPI cards and pie chart based on region selection
@app_kpi_pie.callback(
    [Output('kpi-cards-output', 'children'),
     Output('category-sales-pie-chart', 'figure')],
    [Input('region-dropdown-kpi', 'value')]
)
def update_dashboard(selected_region):
    if selected_region == 'All Regions' or selected_region is None:
        filtered_df = df_superstore
    else:
        filtered_df = df_superstore[df_superstore['Region'] == selected_region]

    # Calculate KPIs
    total_sales = filtered_df['Sales'].sum()
    total_profit = filtered_df['Profit'].sum()
    average_discount = filtered_df['Discount'].mean() * 100 # as percentage

    kpi_cards = [
        html.Div([
            html.H4("Total Sales"),
            html.P(f"${total_sales:,.2f}", style={'fontSize': '24px', 'fontWeight': 'bold'})
        ], style={'textAlign': 'center', 'border': '1px solid #ddd', 'padding': '20px', 'borderRadius': '5px', 'width': '30%'}),
        html.Div([
            html.H4("Total Profit"),
            html.P(f"${total_profit:,.2f}", style={'fontSize': '24px', 'fontWeight': 'bold'})
        ], style={'textAlign': 'center', 'border': '1px solid #ddd', 'padding': '20px', 'borderRadius': '5px', 'width': '30%'}),
        html.Div([
            html.H4("Avg. Discount"),
            html.P(f"{average_discount:.2f}%", style={'fontSize': '24px', 'fontWeight': 'bold'})
        ], style={'textAlign': 'center', 'border': '1px solid #ddd', 'padding': '20px', 'borderRadius': '5px', 'width': '30%'}),
    ]

    # Prepare data for pie chart
    df_category_sales = filtered_df.groupby('Category')['Sales'].sum().reset_index()

    if df_category_sales.empty:
        pie_fig = px.pie(title="No data available for the selected filters.")
    else:
        pie_fig = px.pie(
            df_category_sales,
            values='Sales',
            names='Category',
            title=f'Sales Distribution by Category in {selected_region if selected_region else "All Regions"}',
            hole=0.3, # Creates a donut chart
            color_discrete_sequence=px.colors.sequential.RdBu # Choose a color sequence
        )
        pie_fig.update_traces(textposition='inside', textinfo='percent+label')
        pie_fig.update_layout(showlegend=True)

    return kpi_cards, pie_fig

# Run the app (for Colab, use mode='inline')
app_kpi_pie.run(mode='inline')
```
